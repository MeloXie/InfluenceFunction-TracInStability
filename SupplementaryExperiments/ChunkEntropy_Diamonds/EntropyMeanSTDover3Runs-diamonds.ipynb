```python
import pandas as pd
import numpy as np

# Read the five runs
runs = [
    pd.read_csv("Entropy_Run_1_diamonds.csv"),
    pd.read_csv("Entropy_Run_2_diamonds.csv"),
    pd.read_csv("Entropy_Run_3_diamonds.csv"),
]

# Top K (same for every run)
k = runs[0]["Top K"]

# Stack the entropy values
foif = np.stack([df["FOIF Entropy"].values for df in runs])
tracin = np.stack([df["TracIn Entropy"].values for df in runs])

# Mean and standard deviation
foif_mean = foif.mean(axis=0)
foif_std = foif.std(axis=0)

tracin_mean = tracin.mean(axis=0)
tracin_std = tracin.std(axis=0)

# Final table
result = pd.DataFrame({
    "Top K": k,
    "FOIF": [
        f"{m:.4f} ± {s:.4f}"
        for m, s in zip(foif_mean, foif_std)
    ],
    "TracIn": [
        f"{m:.4f} ± {s:.4f}"
        for m, s in zip(tracin_mean, tracin_std)
    ]
})

print(result)
```

```python
import pandas as pd
import numpy as np

runs = [
    pd.read_csv("Entropy_vs_chunks_Run_1_diamonds.csv"),
    pd.read_csv("Entropy_vs_chunks_Run_2_diamonds.csv"),
    pd.read_csv("Entropy_vs_chunks_Run_3_diamonds.csv"),
]

# Merge all runs
all_runs = pd.concat(runs)

# Compute mean and std for every (n, K)
summary = (
    all_runs
    .groupby(["n", "K"])
    .agg({
        "FOIF Entropy": ["mean", "std"],
        "TracIn Entropy": ["mean", "std"]
    })
)

summary.columns = [
    "FOIF Mean",
    "FOIF Std",
    "TracIn Mean",
    "TracIn Std"
]

summary = summary.reset_index()


summary["FOIF"] = (
    summary["FOIF Mean"].map("{:.4f}".format)
    + " ± "
    + summary["FOIF Std"].map("{:.4f}".format)
)

summary["TracIn"] = (
    summary["TracIn Mean"].map("{:.4f}".format)
    + " ± "
    + summary["TracIn Std"].map("{:.4f}".format)
)

paper_table = summary[["n", "K", "FOIF", "TracIn"]]

print(paper_table)
```
