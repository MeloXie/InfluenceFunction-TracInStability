```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
IF_df = pd.read_csv("IF_Train_Set_Diamonds_Neighbor.csv")
TC_df = pd.read_csv("TC_Train_Set_Diamonds_Neighbor.csv")
```

```python
summary_if = pd.read_csv("summary_if_dia.csv")
summary_tc = pd.read_csv("summary_tc_dia.csv")
```

```python
max_if = IF_df["Score"].max()
min_if = IF_df["Score"].min()
max_difference_if = np.abs(max_if - min_if)

summary_if["norm_avg_inf_diff"] = (
    summary_if["avg_inf_diff"] / max_difference_if
)
```

```python
max_tc = TC_df["Score"].max()
min_tc = TC_df["Score"].min()
max_difference_tc = np.abs(max_tc - min_tc)

summary_tc["norm_avg_inf_diff"] = (
    summary_tc["avg_inf_diff"] / max_difference_tc
)
```

```python
labels = [
    "$(0.00,0.05]$",
    "$(0.05,0.10]$",
    "$(0.10,0.20]$",
    "$(0.20,0.30]$",
    "$(0.30,0.40]$"
]
```

```python
plt.figure(figsize=(7,4))

plt.plot(
    labels,
    summary_if["norm_avg_inf_diff"],
    marker="o",
    linewidth=2.5,
    label="FOIF"
)

plt.plot(
    labels,
    summary_tc["norm_avg_inf_diff"],
    marker="s",
    linewidth=2.5,
    label="TracIn"
)




plt.ylabel(
    r"Normalised influence difference",
    fontsize=14
)

plt.xlabel(
    r"Distance band  ($ \times m_d$)",
    fontsize=18
)

plt.xticks(
    fontsize=14
)

plt.yticks(
    fontsize=14
)


plt.grid(True, alpha=0.3)
plt.legend(fontsize=14)

plt.savefig(
    "Neighbour_Radius_new_dia.png",
    dpi=600,
    bbox_inches="tight"
)


plt.show()
```
